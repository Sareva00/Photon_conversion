```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import tol_colors as tc

import tools
import my_plot as mplt
```

```python
colors = mplt.set_colors(3, palette="colorblind")
```

```python
def extract_custom_state_from_files(folder_path, ell, ell_max):
    """Extract custom transfer function data from files in a folder """
    folder = Path(folder_path) / "FH" / "aniso_system" 
    print(f"Looking for files in: {folder}")
    file_path = (next(folder.glob(f"custom_state.dat_lmax{ell_max}_adiabatic.dat")))
    
    theta = [[] for _ in range(ell+1)]
    z = []
    
    data = np.loadtxt(file_path, comments='#')  
    z.extend(data[:, 0])  # First column is redshift
    for i in range(ell+1):
        theta[i].extend(data[:, 1+i])   # All theta values from column 1 and multiples of it
        
    return z, np.array(theta)
```

# First case: tight coupling with no damping or polarisation
$\Theta_{\ell} = 0$ for $\ell>1$ 

all the polarisation states are set to zero

```python
z , theta0 = extract_custom_state_from_files("/Users/user/Documents/Data/Dark-photons/Tight_coupling/tight-coupling_tf0", 2, 5)
```

```python
z , theta1 = extract_custom_state_from_files("/Users/user/Documents/Data/Dark-photons/Tight_coupling/tight-coupling_tf1", 1, 1)
#a , theta1 = extract_custom_state_from_files("/Users/user/Documents/Data/Dark-photons/Tight_coupling/tight-coupling_tf1", 1, 5)
```

```python

fig, ax = plt.subplots(1, 1, figsize=mplt.set_size(mplt.width))

ax.plot(z, theta1[0], color=colors[0], alpha=0.7, linestyle="-", label="$\\Theta_0$")  # Assuming theta1[0] corresponds to theta_0
ax.plot(z, theta0[0], color=colors[0], alpha=0.7, linestyle="--", label="$\\Theta_0$ (FS)")  
ax.plot(z, theta1[1], color=colors[1], alpha=0.7, linestyle="-", label="$\\Theta_1$")
ax.plot(z, theta0[1], color=colors[1], alpha=0.7, linestyle="--", label="$\\Theta_1$ (FS)")

mplt.set_ticks(ax)
mplt.set_labels(ax, x_label="z", x_log=True, y_label="Transfer Function", y_log=False, title=None)


ax.legend()
plt.show()
```
